```python
# import libraries (you may add additional imports but you may not have to)
import numpy as np
import pandas as pd
from scipy.sparse import csr_matrix
from sklearn.neighbors import NearestNeighbors
import matplotlib.pyplot as plt

def choose_column(df, potential_names, df_name):
    for name in potential_names:
        if name in df.columns:
            print(f"Column '{name}' selected for {df_name}.")
            return name
    raise ValueError(f"Could not find any of the specified columns {potential_names} in dataframe '{df_name}'.")
```

```python
# get data files
!wget https://cdn.freecodecamp.org/project-data/books/book-crossings.zip

!unzip book-crossings.zip

books_filename = 'BX-Books.csv'
ratings_filename = 'BX-Book-Ratings.csv'
```

```python
# import csv data into dataframes
df_books = pd.read_csv(
    books_filename,
    encoding = "ISO-8859-1",
    sep=";",
    header=0,
    names=['isbn', 'title', 'author'],
    usecols=['isbn', 'title', 'author'],
    dtype={'isbn': 'str', 'title': 'str', 'author': 'str'})

df_ratings = pd.read_csv(
    ratings_filename,
    encoding = "ISO-8859-1",
    sep=";",
    header=0,
    names=['user', 'isbn', 'rating'],
    usecols=['user', 'isbn', 'rating'],
    dtype={'user': 'int32', 'isbn': 'str', 'rating': 'float32'})
```

```python
book_rating_column_name = choose_column(df_ratings, ['Book-Rating', 'rating', 'book_rating', 'Rating'], 'df_ratings')
user_column_name = choose_column(df_ratings, ['User-ID', 'UserID', 'user_id', 'user'], 'df_ratings')
bookid_col_ratings = choose_column(df_ratings, ['ISBN', 'isbn', 'Book-ID', 'BookID', 'book_id'], 'df_ratings')
bookid_col_books   = choose_column(df_books,   ['ISBN', 'isbn', 'book_id', 'Book-ID', 'BookID'], 'df_books')
title_col = choose_column(df_books, ['Book-Title', 'Title', 'title', 'book_title'], 'df_books')
ratings = df_ratings.copy()
books = df_books.copy()
ratings = ratings.merge(books[[bookid_col_books, title_col]],
                        left_on=bookid_col_ratings,
                        right_on=bookid_col_books,
                        how='left')
ratings = ratings.rename(columns={book_rating_column_name: 'rating', user_column_name: 'user_id', title_col: 'title'})
ratings['rating'] = pd.to_numeric(ratings['rating'], errors='coerce')
ratings = ratings.dropna(subset=['rating'])
```

```python
# Filter out books with less than 100 ratings
book_rating_counts = ratings['title'].value_counts()
rats = ratings[ratings['title'].isin(book_rating_counts[book_rating_counts >= 100].index)]

# Filter out users with less than 200 ratings
user_rating_counts = rats['user_id'].value_counts()
rats = rats[rats['user_id'].isin(user_rating_counts[user_rating_counts >= 200].index)]

# Create pivot table
book_pivot = rats.pivot_table(index='user_id', columns='title', values='rating').fillna(0)

# Convert pivot table to sparse matrix
book_sparse_matrix = csr_matrix(book_pivot.values)

# Fit NearestNeighbors model on the transpose for item-based recommendations
model = NearestNeighbors(metric='cosine', algorithm='brute')
model.fit(book_sparse_matrix.T)
```

```python
# function to return recommended books - this will be tested
def get_recommends(book = ""):
  try:
    book_index = book_pivot.columns.get_loc(book)
  except KeyError:
    # If the book is not in our filtered dataset, we can't recommend.
    # This also handles cases where a book has less than 200 ratings.
    print(f"Book '{book}' not found in the dataset used for recommendations. Please try another book.")
    return []

  distances, indices = model.kneighbors(book_sparse_matrix[book_index], n_neighbors=6)

  recommended_books = [book, []]
  for i in range(1, len(distances.flatten())):
      title = book_pivot.columns[indices.flatten()[i]]
      distance = distances.flatten()[i]
      recommended_books[1].append([title, distance])

  return recommended_books
```

```python
# function to return recommended books - this will be tested
def get_recommends(book = ""):
  try:
    book_index = book_pivot.columns.get_loc(book)
  except KeyError:
    # If the book is not in our filtered dataset, we can't recommend.
    # This also handles cases where a book has less than 200 ratings.
    print(f"Book '{book}' not found in the dataset used for recommendations. Please try another book.")
    return []

  # Query the transposed sparse matrix using the book's column index as a row index
  distances, indices = model.kneighbors(book_sparse_matrix.T[book_index], n_neighbors=6)

  recommended_books = [book, []]
  for i in range(1, len(distances.flatten())):
      # The indices returned are for the rows of the matrix the model was fitted on (book_sparse_matrix.T)
      # These rows correspond to the columns of the original book_pivot.
      title = book_pivot.columns[indices.flatten()[i]]
      distance = distances.flatten()[i]
      recommended_books[1].append([title, distance])

  return recommended_books
```
